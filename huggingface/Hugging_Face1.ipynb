```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
```

```python
result = classifier("I love machine learning!")
result
```

```python
texts = [
    "This lesson was amazing!",
    "I am very disappointed with my exam score.",
    "I think I need more practice in math."
]

for t in texts:
    print(t, "->", classifier(t))
```
